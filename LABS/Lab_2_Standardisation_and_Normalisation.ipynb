```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Aditya Narayan Ray")

# Load and clean dataset from Lab 1 to prepare df_clean
df = pd.read_csv("hotel_booking.csv")
df_clean = df.drop(columns=['company'])
df_clean['agent'] = df_clean['agent'].fillna(0)
df_clean['country'] = df_clean['country'].fillna(df_clean['country'].mode()[0])
df_clean['children'] = df_clean['children'].fillna(0)
guest_filter = (df_clean['adults'] + df_clean['children'] + df_clean['babies']) > 0
df_clean = df_clean[guest_filter]
```

```python
# Create a copy for standardization
df_standardized = df_clean.copy()

# Standardize lead_time
lead_time_mean = df_standardized['lead_time'].mean()
lead_time_std = df_standardized['lead_time'].std()
df_standardized['lead_time'] = (df_standardized['lead_time'] - lead_time_mean) / lead_time_std

# Standardize adr
adr_mean = df_standardized['adr'].mean()
adr_std = df_standardized['adr'].std()
df_standardized['adr'] = (df_standardized['adr'] - adr_mean) / adr_std

print(df_standardized[['lead_time', 'adr']].head())
```

```python
comparison = df_clean[['lead_time', 'adr']].copy()
comparison['lead_time_Standardized'] = df_standardized['lead_time']
comparison['adr_Standardized'] = df_standardized['adr']
print(comparison.head())
```

```python
# Plot Original vs Standardized lead_time
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.hist(df_clean['lead_time'], bins=20)
plt.title("Original Lead Time")
plt.subplot(1, 2, 2)
plt.hist(df_standardized['lead_time'], bins=20)
plt.title("Standardized Lead Time")
plt.tight_layout()
plt.show()
```

```python
# Create a copy for normalization
df_normalized = df_clean.copy()

# Normalize lead_time
lead_time_min = df_normalized['lead_time'].min()
lead_time_max = df_normalized['lead_time'].max()
df_normalized['lead_time'] = (df_normalized['lead_time'] - lead_time_min) / (lead_time_max - lead_time_min)

# Normalize adr
adr_min = df_normalized['adr'].min()
adr_max = df_normalized['adr'].max()
df_normalized['adr'] = (df_normalized['adr'] - adr_min) / (adr_max - adr_min)

print(df_normalized[['lead_time', 'adr']].head())
```

```python
# Create comparison dataframe
comparison = df_clean[['lead_time', 'adr']].copy()
comparison['lead_time_Normalized'] = df_normalized['lead_time']
comparison['adr_Normalized'] = df_normalized['adr']

print(comparison.head())
```

```python
# Plot Original vs Normalized lead_time
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.hist(df_clean['lead_time'], bins=20)
plt.title("Original Lead Time")
plt.subplot(1, 2, 2)
plt.hist(df_normalized['lead_time'], bins=20)
plt.title("Normalized Lead Time")
plt.tight_layout()
plt.show()
```
