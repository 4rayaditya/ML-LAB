```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Aditya Narayan Ray")
```

```python
df = pd.read_csv("hotel_booking.csv")
print(df.select_dtypes(include=np.number).max())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols].hist(figsize=(20, 15), bins=20, edgecolor='black')
plt.suptitle("Distribution of Numerical Variables", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
correlation = df[numeric_cols].corr()

plt.figure(figsize=(16, 12))
sns.heatmap(
    correlation,
    annot=False, # Set to False because 20x20 text annotations will overlap
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Select a targeted subset of numerical features for the pairplot
pairplot_cols = ['lead_time', 'adr', 'days_in_waiting_list', 'total_of_special_requests', 'is_canceled']

sns.pairplot(
    df[pairplot_cols],
    diag_kind="hist",
    hue="is_canceled" # Added hue to separate distributions by cancellation
)
plt.suptitle("Pair Plot of Key Numerical Variables", y=1.02)
plt.show()
```
