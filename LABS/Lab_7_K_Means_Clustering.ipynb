```python
import os
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

data = load_iris()
X = pd.DataFrame(data.data, columns=data.feature_names)
print("Author: Aditya Narayan Ray")
print(f"Dataset loaded: {X.shape[0]} rows and {X.shape[1]} features.")
```

```python
inertias = []
K = range(1, 10)
for k in K:
    kmeanModel = KMeans(n_clusters=k, random_state=42, n_init='auto')
    kmeanModel.fit(X)
    inertias.append(kmeanModel.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K, inertias, 'ko-', linewidth=2, markersize=8)
plt.title('Elbow Method For Optimal k', fontsize=14, pad=15)
plt.xlabel('Number of Clusters (k)', fontsize=12)
plt.ylabel('Inertia (Within-Cluster Sum of Squares)', fontsize=12)
plt.grid(True, linestyle=':', color='dimgrey', alpha=0.7)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init='auto')
X['Cluster'] = kmeans.fit_predict(X)

plt.figure(figsize=(9, 6))
markers = ['o', 's', '^']
colors = ['white', 'silver', 'dimgrey']

for i in range(3):
    cluster_data = X[X['Cluster'] == i]
    plt.scatter(cluster_data.iloc[:, 0], cluster_data.iloc[:, 1], c=colors[i], marker=markers[i],
                label=f'Cluster {i+1}', s=60, edgecolors='black', alpha=0.8)

centroids = kmeans.cluster_centers_
plt.scatter(centroids[:, 0], centroids[:, 1], c='black', marker='X', s=300, label='Centroids')
plt.title('K-Means Clustering Results', fontsize=14, pad=15)
plt.xlabel(X.columns[0], fontsize=12)
plt.ylabel(X.columns[1], fontsize=12)
plt.legend(frameon=False, loc='upper right')
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```
